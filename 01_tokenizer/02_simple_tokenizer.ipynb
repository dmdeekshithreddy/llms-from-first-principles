#### Step 1: Read the input text

```python
from pathlib import Path


text_file = Path("../datasets/the-verdict.txt")

with open(text_file, "r", encoding="utf-8") as f:
    raw_text = f.read()

print("Total characters:", len(raw_text))
print(raw_text[:100])
```

#### Step 2: Split the text into tokens using regex

```python
# Check the non alphanumeric characters in the text
import re

non_alphanumeric = re.findall(r'[^a-zA-Z0-9\s]', raw_text)
print("Non-alphanumeric characters:", set(non_alphanumeric))
```
